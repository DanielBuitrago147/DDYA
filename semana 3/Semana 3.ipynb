 Ejercicio: Organización y búsqueda de productos en una tienda
Una tienda maneja una lista con los *códigos numéricos de sus productos**. Como los productos se registran a medida que llegan, los códigos inicialmente se encuentran desordenados.

El encargado necesita desarrollar un programa que permita:

1. Registrar una cantidad n de productos.
2. Ordenar sus códigos utilizando *Merge Sort**.
3. Buscar un código específico utilizando *Búsqueda Binaria**.

 Ejemplo

[105, 32, 78, 15, 120, 50]

 Punto 1. Ordenar los productos con Merge Sort
Utilizando el algoritmo *Merge Sort**, organice los códigos de los productos de menor a mayor.

Resultado esperado:

[15, 32, 50, 78, 105, 120]Recuerde que Merge Sort utiliza la estrategia de *Dividir y Conquistar**:

- *Dividir:** separar la lista en dos partes.
- *Conquistar:** ordenar cada mitad.
- *Combinar:** unir nuevamente las mitades de forma ordenada.

## 7) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación: El programa ordena los códigos con merge_sort para dejarlos de menor a mayor. Después usa búsqueda binaria, que revisa el elemento del centro y descarta la mitad que no puede contener el código. Esto hace más eficiente la búsqueda en listas grandes.
- Función T[n]: Registro de productos: O(n), ya que se deben ingresar los n códigos.
Ordenamiento con Merge Sort: O(nlogn), tanto en el mejor, promedio como peor caso.
Búsqueda Binaria: O(logn) en el caso promedio y peor caso, y O(1) en el mejor caso.

Por lo tanto, la complejidad total del programa es O(nlogn), debido a que Merge Sort es el proceso que tiene el mayor costo.

En cuanto al espacio, se requiere O(n) para almacenar los códigos de los productos.

```python
def merge_sort(lista):
    if len(lista) <= 1:
        return lista

    mitad = len(lista) // 2

    izquierda = merge_sort(lista[:mitad])
    derecha = merge_sort(lista[mitad:])

    return merge(izquierda, derecha)


def merge(izquierda, derecha):
    ordenada = []

    while izquierda and derecha:
        if izquierda[0] < derecha[0]:
            ordenada.append(izquierda.pop(0))
        else:
            ordenada.append(derecha.pop(0))

    return ordenada + izquierda + derecha
def busqueda_binaria(lista, objetivo):
    izquierda = 0
    derecha = len(lista) - 1

    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2

        if lista[medio] == objetivo:
            return medio
        elif lista[medio] < objetivo:
            izquierda = medio + 1
        else:
            derecha = medio - 1

    return -1


def main():
    entrada = input("Ingrese numeros separados por espacios: ")
    lista = []

    for numero in entrada.split():
        lista.append(int(numero))

    print("Lista original:", lista)
    print("Lista ordenada:", merge_sort(lista))

    objetivo = int(input("Ingrese el numero a buscar: "))
    indice = busqueda_binaria(merge_sort(lista), objetivo)
    if indice != -1:
        print(f"El numero {objetivo} se encuentra en la posicion {indice}.")
    else:
        print(f"El numero {objetivo} no se encuentra en la lista.")
main()
```
