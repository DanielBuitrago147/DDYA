## 1) Descripción del problema **[1 pts]**
**Ejercicio: Inventario de productos utilizando Tabla Hash**


Contexto
Una tienda necesita mejorar la forma en que consulta los productos de su inventario.

Actualmente tiene muchos productos registrados y recorrer toda una lista cada vez que necesita buscar uno resulta poco eficiente.

Para solucionar este problema se deberá implementar una Tabla Hash, utilizando como llave el código numérico del producto.

Cada producto tendrá:

Código Nombre Precio Cantidad disponible
Por ejemplo:

101 - Arroz       - $4.500  - 20 unidades 205 - Leche       - $3.800  - 15 unidades 312 - Café        - $12.000 - 10 unidades 423 - Pan         - $2.500  - 30 unidades
Punto 1. Crear la Tabla Hash
El usuario deberá indicar el tamaño m de la tabla.

Ejemplo:

Ingrese el tamaño de la tabla hash: 7
Inicialmente la tabla estará vacía:

Índice     Producto 0          Vacío 1          Vacío 2          Vacío 3          Vacío 4          Vacío 5          Vacío 6          Vacío
Para determinar la posición de cada producto se deberá utilizar:

h(código) = código % m
Donde:

código es la llave del producto.

m es el tamaño de la tabla.

Punto 2. Insertar productos
Permita al usuario registrar una cantidad n de productos.

Por ejemplo:

¿Cuántos productos desea registrar?: 5 Código: 101 Nombre: Arroz Precio: 4500 Cantidad: 20
Para cada producto deberá calcularse su posición utilizando la función hash.

Si:

m = 7 código = 101
entonces:

101 % 7 = 3
Por lo tanto, el producto deberá almacenarse inicialmente en la posición 3.

Después de insertar todos los productos, muestre el estado completo de la tabla.

Punto 3. Manejo de colisiones
Puede ocurrir que dos productos generen el mismo índice.

Por ejemplo:

Código 101: 101 % 7 = 3 Código 108: 108 % 7 = 3
Existe una colisión porque ambos productos intentan ocupar la posición 3.

Las colisiones deberán resolverse utilizando encadenamiento.

Por ejemplo:

Índice 3: 101 - Arroz      ↓ 108 - Huevos      ↓ 115 - Azúcar
Cada posición de la tabla podrá contener varios productos cuando exista una colisión.

Punto 4. Buscar un producto
El usuario deberá ingresar el código del producto que desea consultar.

Ejemplo:

Ingrese el código del producto: 108
El programa deberá:

Calcular su posición mediante la función hash.

Acceder directamente a esa posición.

Buscar el producto dentro de la posición correspondiente.

Si existe:

Producto encontrado Código: 108 Nombre: Huevos Precio: $16.000 Cantidad disponible: 12
Si no existe:

El producto no está registrado.
No se deberá recorrer toda la Tabla Hash para realizar la búsqueda.

Punto 5. Eliminar un producto
Permita eliminar un producto utilizando su código.

Ejemplo:

Código del producto que desea eliminar: 108

Antes:

Índice 3: 101 - Arroz 108 - Huevos 115 - Azúcar
Después:

Índice 3: 101 - Arroz 115 - Azúcar
Si el código no existe, deberá mostrar:

No se encontró el producto.
Punto 6. Actualizar inventario
Permita buscar un producto por su código y modificar su cantidad disponible.

Ejemplo:

Código: 101 Producto: Arroz Cantidad actual: 20 Nueva cantidad: 35
Resultado:

Inventario actualizado correctamente. Arroz - 35 unidades
La búsqueda del producto deberá realizarse utilizando la Tabla Hash.

### 5) Análisis de complejidad **[15 pts]**

Justificación

Mejor caso:

Se presenta cuando los productos se distribuyen adecuadamente entre las posiciones de la tabla y no existen colisiones importantes.

Por ejemplo, al buscar un producto, la función hash calcula su posición y el producto se encuentra inmediatamente.

Por eso, la búsqueda puede tener una complejidad de \(O(1)\).

Peor caso:

Se presenta cuando muchos productos generan la misma posición hash.

En ese escenario, todos quedan almacenados en una misma cadena y el programa debe recorrerla para encontrar el producto solicitado.

Si existen \(n\) productos en esa cadena, la búsqueda puede llegar a tener una complejidad de \(O(n)\).


Mejor caso:
\[
T(n)=c
\]

Donde \(c\) representa el tiempo constante necesario para calcular la posición y encontrar el producto inmediatamente.

Por lo tanto:
\[
T(n)=O(1)
\]

Peor caso:
\[
T(n)=an+b
\]

Donde:
- \(a\) representa el tiempo necesario para revisar cada elemento.
- \(n\) representa la cantidad de productos que deben revisarse.
- \(b\) representa las operaciones constantes adicionales.

Por lo tanto:
\[
T(n)=O(n)
\]

## 6) Código funcional **[15 pts]**


```python
def crear_tabla(tamano):

    tabla = []

    for i in range(tamano):
        tabla.append([])

    return tabla

def funcion_hash(codigo, tamano):

    return codigo % tamano


def insertar_producto(tabla, tamano):

    codigo = int(input("Código: "))
    nombre = input("Nombre: ")
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad disponible: "))

    indice = funcion_hash(codigo, tamano)

    for producto in tabla[indice]:

        if producto["codigo"] == codigo:
            print("Ese código ya existe.")
            return

    nuevo_producto = {
        "codigo": codigo,
        "nombre": nombre,
        "precio": precio,
        "cantidad": cantidad
    }

    tabla[indice].append(nuevo_producto)

    print("Producto registrado correctamente.")


def registrar_productos(tabla, tamano):

    cantidad = int(input("¿Cuántos productos desea registrar?: "))

    for i in range(cantidad):

        print("\nProducto", i + 1)

        insertar_producto(tabla, tamano)

def mostrar_tabla(tabla):

    print("\n========================================")
    print("              TABLA HASH")
    print("========================================")

    for i in range(len(tabla)):

        print(i, end=" -> ")

        if len(tabla[i]) == 0:

            print("Vacío")

        else:

            for producto in tabla[i]:

                print(
                    f'{producto["codigo"]} - {producto["nombre"]} '
                    f'(${producto["precio"]}) '
                    f'Cantidad: {producto["cantidad"]}',
                    end="   "
                )

            print()

def buscar_producto(tabla, tamano):

    codigo = int(input("Ingrese el código del producto: "))

    indice = funcion_hash(codigo, tamano)

    encontrado = False

    for producto in tabla[indice]:

        if producto["codigo"] == codigo:

            print("\nProducto encontrado")
            print("--------------------------")
            print("Código:", producto["codigo"])
            print("Nombre:", producto["nombre"])
            print("Precio: $", producto["precio"])
            print("Cantidad disponible:", producto["cantidad"])

            encontrado = True
            break


    if encontrado == False:

        print("El producto no está registrado.")

def eliminar_producto(tabla, tamano):

    codigo = int(input("Código del producto que desea eliminar: "))

    indice = funcion_hash(codigo, tamano)


    for producto in tabla[indice]:

        if producto["codigo"] == codigo:

            tabla[indice].remove(producto)

            print("Producto eliminado correctamente.")

            return


    print("No se encontró el producto.")


def actualizar_inventario(tabla, tamano):

    codigo = int(input("Ingrese el código del producto: "))

    indice = funcion_hash(codigo, tamano)


    for producto in tabla[indice]:

        if producto["codigo"] == codigo:


            print("\nProducto encontrado")
            print("Producto:", producto["nombre"])
            print("Cantidad actual:", producto["cantidad"])


            nueva_cantidad = int(
                input("Nueva cantidad: ")
            )


            producto["cantidad"] = nueva_cantidad


            print("\nInventario actualizado correctamente.")
            print(
                producto["nombre"],
                "-",
                producto["cantidad"],
                "unidades"
            )

            return



    print("No se encontró el producto.")



def main():

    tabla = None
    tamano = 0


    while True:

        print("\n========================================")
        print("     INVENTARIO DE PRODUCTOS - HASH")
        print("========================================")

        print("1. Crear Tabla Hash")
        print("2. Registrar productos")
        print("3. Mostrar Tabla Hash")
        print("4. Buscar producto")
        print("5. Eliminar producto")
        print("6. Actualizar inventario")
        print("7. Salir")


        opcion = input("\nSeleccione una opción: ")

        if opcion == "1":

            tamano = int(input("Ingrese el tamaño de la tabla hash: "))

            tabla = crear_tabla(tamano)

            print("Tabla Hash creada correctamente.")


        elif opcion == "2":

            if tabla == None:

                print("Primero debe crear la Tabla Hash.")

            else:

                registrar_productos(tabla, tamano)

        elif opcion == "3":

            if tabla == None:

                print("Primero debe crear la Tabla Hash.")

            else:

                mostrar_tabla(tabla)

        elif opcion == "4":

            if tabla == None:

                print("Primero debe crear la Tabla Hash.")

            else:

                buscar_producto(tabla, tamano)

        elif opcion == "5":

            if tabla == None:

                print("Primero debe crear la Tabla Hash.")

            else:

                eliminar_producto(tabla, tamano)

        elif opcion == "6":

            if tabla == None:

                print("Primero debe crear la Tabla Hash.")

            else:

                actualizar_inventario(tabla, tamano)

        elif opcion == "7":

            print("Programa finalizado.")

            break



        else:

            print("Opción inválida.")


main()
```
